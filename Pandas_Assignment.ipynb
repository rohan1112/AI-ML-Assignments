```python
# 25.	Convert the hours logged column to a number.
df['hours_logged'] = pd.to_numeric(df['hours_logged'],errors='coerce')

print("\nConverting hours loggeg column to the numeric type")
```

```python
# 26.	Convert the billing rate column to a number. Remove the $ sign first.

df['billing_rate_(usd)'] = (
    df['billing_rate_(usd)']
    .astype(str)
    .str.replace(r'[^\d.]', '', regex=True)
)

df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'], errors='coerce')

print(df['billing_rate_(usd)'])

print("\nConverting billing rate column to the numeric type and removing the $ and displayng the result")

```

```python
# 27.	Convert the CSAT score column to a number.

df['csat_score'] = pd.to_numeric(df['csat_score'],errors='coerce')

print(df['csat_score'])
print("\nConverting CSAT score column to the numeric type")
```

```python
# 28.	Print the average, the smallest and the largest value of hours logged.

print('Average Hours Logged:', df['hours_logged'].mean())
print('Smallest Hours Logged:', df['hours_logged'].min())
print('Largest Hours Logged:', df['hours_logged'].max())

print("\nDispaying avg min and max value from hour logged column")
```

```python
# 29.	Sort the table by hours logged from highest to lowest and show the top 10 rows.
df['hours_logged'].sort_values(ascending=False).head(10)

print("\nDisplaying hours logged column from highes to lowest value")

```

```python
# 30.	Show the 5 tickets with the highest billable amount.

df['billable_amount_usd'] = pd.to_numeric(df['billable_amount_usd'], errors='coerce')

df.sort_values(by='billable_amount_usd', ascending=False).head(5)

print("\nDisplaying top 5 highest billable amounts")
```

```python
# 31.	Remove the extra spaces from the client name column.

df['client_name'] = df['client_name'].str.strip()

print(df['client_name'])

# str.strip() removes extra spaces from the beginning and end of each text value.
print("\nClient names after removing extra spaces")
```

```python
# 32.	Change the client name column to title case.

df['client_name'] = df['client_name'].str.title()

print(df['client_name'])

print("\nClient names title case appiled")
```

```python
# 33.	Convert the ticket created column to a date.

print(df['ticket_created_at'])
df['ticket_created_at'] = pd.to_datetime(
    df['ticket_created_at'],
    format='mixed',
    errors='coerce'
)
print(df['ticket_created_at'])

print("\nConverting ticket created column to data column which was in mixed format")

```

```python
# 34.	Add a new column that holds the year of the ticket.

df['ticket_year'] = df['ticket_created_at'].dt.year

print(df['ticket_year'])

# taking year from the dataframe and saving in ticker_year column
print("\ntaking year from the dataframe and saving in ticker_year column")
```

```python
# 35.	Add a new column that holds the month of the ticket.

df['ticket_month'] = df['ticket_created_at'].dt.month

print(df['ticket_month'])

print("\ntaking month  from the dataframe and saving in ticker_month column")
```

```python
# 36.	Count how many tickets were created in each year.

df.groupby('ticket_year').size()

print("\nTickets created for each year")
```

```python
# 37.	Add a new column called total equal to billing rate multiplied by hours logged.
df.columns

df['total_equal'] = df['billable_amount_usd'] * df['hours_logged']

print(df['total_equal'])


print("\nDisplaying total hours worked")

```

```python
# 38.	Fill the missing values in the CSAT score column with the average CSAT score.
df['csat_score'] = df['csat_score'].fillna(df['csat_score'].mean())

print(df['csat_score'])
print("\nFilling the missing values with avg CSAt score")
```

```python
# 39.	Drop the rows where the client name is missing.
df = df.dropna(subset=['client_name'])
print("\nDroped the rows where client name was missing")
```
